//========================TASK 1=====================//

```python
import pandas as pd

df = pd.read_excel("flipkart_sales_data (1).xlsx")

# 1. Display first five rows
print("\nFIRST 5 ROWS:")
display(df.head())

# 2. Display last five rows
print("\nLAST 5 ROWS:")
display(df.tail())

# 3. Number of rows and columns
print("\nNUMBER OF ROWS AND COLUMNS:")
print(df.shape)

# 4. Display all column names
print("\nCOLUMN NAMES:")
print(df.columns.tolist())

# 5. Check data types of columns
print("\nDATA TYPES:")
print(df.dtypes)

# 6. Basic statistical information
print("\nBASIC STATISTICAL INFORMATION:")
display(df.describe())
```

//=========================TASK 2=======================//

```python
#1. Find missing values and their percentage

# Number of missing values in each column
missing_count = df.isnull().sum()

# Percentage of missing values in each column
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': missing_percentage
})

print("MISSING VALUE SUMMARY:")
display(missing_summary)

#2. Display only columns that contain missing values

print("\nCOLUMNS WITH MISSING VALUES:")
display(missing_summary[missing_summary['Missing Values'] > 0])

#3. Handle the missing values

#Numerical columns → replace missing values with the median.
#Categorical columns → replace missing values with the mode.

# Separate numerical and categorical columns
numeric_columns = df.select_dtypes(include=['number']).columns
categorical_columns = df.select_dtypes(include=['object', 'category']).columns


# Fill missing numerical values with median
for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())


# Fill missing categorical values with mode
for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

#4. Verify after handling missing values

print("\nMISSING VALUES AFTER CLEANING:")
missing_after = df.isnull().sum()
display(missing_after[missing_after > 0])
print("\nTotal missing values in the dataset:", df.isnull().sum().sum())
```

Explanation:          

Missing Value Handling:

First, the number and percentage of missing values were calculated for each column. For numerical columns, missing values were replaced with the median value because the median is less affected by extreme values or outliers than the mean. For categorical columns, missing values were replaced with the mode, which represents the most frequently occurring category. This approach was selected instead of removing entire records because deleting records could result in the loss of useful customer and sales information. After applying the cleaning methods, the dataset was checked again to verify that the missing values had been handled successfully.

//===========================TASK 3=========================//

```python
# 1. Find duplicate records
duplicate_count = df.duplicated().sum()
print("TOTAL NUMBER OF DUPLICATE RECORDS:", duplicate_count)

# 2. Display duplicate records
print("\nDUPLICATE RECORDS:")
duplicates = df[df.duplicated(keep=False)]
display(duplicates.head(10))

# 3. Remove duplicate records
df = df.drop_duplicates()
print("\nDuplicate records removed successfully.")

# 4. Verify
remaining_duplicates = df.duplicated().sum()
print("\nNUMBER OF DUPLICATES AFTER REMOVAL:", remaining_duplicates)
print("NUMBER OF ROWS AFTER REMOVING DUPLICATES:", len(df))
```

//=================TASK 4=========================//

```python
print("DATA TYPES BEFORE CLEANING:")
print(df.dtypes)

# Problem 1: Extra spaces in text

text_columns = df.select_dtypes(include=['object']).columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("\nProblem 1: Extra spaces removed from text columns.")

# Problem 2: Inconsistent capitalization

for col in text_columns:
    df[col] = df[col].str.title()

print("Problem 2: Capitalization standardized.")

# Verify the corrected data

print("\nDATA TYPES AFTER CLEANING:")
print(df.dtypes)

print("\nSAMPLE OF CLEANED DATA:")
display(df.head())

print("\nUNIQUE VALUES AFTER CLEANING:")

for col in text_columns:
    print("\nColumn:", col)
    print(df[col].dropna().unique()[:20])
```

Explanation:

Wrong Data Format and Consistency Handling:

The dataset was examined for inconsistent data formats and values. Two consistency problems were addressed: extra spaces in text values and inconsistent capitalization. The str.strip() function was used to remove unnecessary spaces from text values, while str.title() was used to standardize capitalization. These corrections ensure that values representing the same category are stored consistently. The affected columns were then examined again to verify that the formatting was consistent.

//=======================TASK 5=======================//

```python
# Select a numerical column from your dataset
column = 'Sales Amount'

# Calculate Q1 and Q3
Q1 = df[column].quantile(0.25)
Q3 = df[column].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Column:", column)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Identify outliers
outliers = df[
    (df[column] < lower_bound) |
    (df[column] > upper_bound)
]

print("\nNumber of outliers:", len(outliers))

print("\nIDENTIFIED OUTLIERS:")
display(outliers)
```

Explanation:

Outlier Detection and Handling:

The IQR method was used to identify potential outliers in the numerical Price column. First, Q1 and Q3 were calculated, followed by the IQR using IQR = Q3 - Q1. The lower and upper bounds were calculated as Q1 - 1.5 × IQR and Q3 + 1.5 × IQR, respectively. Values outside these bounds were identified as potential outliers. The detected outliers were retained because, in a real-world sales dataset, unusually high or low product prices can represent legitimate products rather than incorrect data. Therefore, removing them could result in the loss of meaningful information.
